```python
def grad(f, x, h=1e-3):
    return (f(x + h) - f(x - h)) / (2 * h)
```

```python
def f_(f, x0, r=0.1, threshold=1e-6):
    x = x0
    step = 0
    while True:
        g = grad(f, x)
        x = x - r * g
        step += 1
        print(f"Time: {step} Current x: {x:.8f}, f(x): {f(x):.8f}, Gradient: {g:.8f}")
        if abs(g) < threshold:
            break
    return x
```

```python
def f_plot(f, x0, r=0.1, threshold=1e-6):
    x = x0
    data = []
    while True:
        g = grad(f, x)
        x = x - r * g
        data.append(g)
        if abs(g) < threshold:
            break
    return data
```

```python
result = f_(lambda x: x**2, x0=10, r=0.1)
```

```python
result_plot_01 = f_plot(lambda x: x**2, x0=10, r=0.1)

import matplotlib.pyplot as plt
plt.plot(result_plot_01)
plt.xlabel('Iteration')
plt.ylabel('Gradient')
plt.title('Gradient Descent Progress')

```
